# step 1 - Create a Simple Dataset

## Task 1 Input generation

```python
import numpy as np
import pandas as pd
```

```python
x=np.linspace(0,10,num=50).reshape(50,1)
x.shape
```

Q) what does x represent


Ans It represents the input variable which is used to predict the output

Q) Why was the range 0–10 chosen?

Ans. This range ensures we see both early rapid change and later saturation, which a
linear model cannot fit well

## Task 1.2 Target generation

```python
noise = np.random.normal(0,0.2,size=(50,1))
y=np.log(x+1) + noise
```

Q) why this target cannot be fit well by a straight line

Ans. Because the logarithm function does not follows a straight line as it has high slope at small x and small slope at large x

# Decide the Model Shape

Q) How many intermediate values (hidden units) do you want?


Ans. 3

Q) Why more than 1?


Ans. A single hidden behaves almost like a linear model thus as our ouput is non linear we are using the more than one hidden units in order to capture more details

Q) why not too many?

Ans. Too many hidden units increase complexity and can lead to overfitting

# Step 3


```python
w1 = np.random.uniform(-1,1,size=(1,3))
b1 = np.zeros((1,3))
w2 = np.random.uniform(-1,1,size=(3,1))
b2 = np.zeros((1,1))
```

Q) what each parameter controls

Weights affect shape. Biases affect position.

# step 4 Defining the Activation Function


## task 4.1 write the function

```python
def activation(z):
  return np.maximum(0,z)
```

## task 4.2 Write it's slope

```python
def activation_slope(z):
  return (z>0).astype(float)
```

Q) why slope matters for learning

Ans. The slope tells the model how to change parameters to reduce error. Without slope, learning cannot happen

# step 5 Forward Pass

```python
# step 1 : Linear Transformation
z1 =  x @ w1 + b1
# Step 2: Apply activation (introduces non-linearity)
h = activation(z1)
# Step 3: Compute final output prediction
y_hat = h @ w2 + b2
```

For z1 = X @ W1 + b1

Computes weighted sum of inputs for hidden neurons.

For h = activation(z1)

Applies non-linear transformation to enable bending.

For y_hat = h @ W2 + b2

Combines hidden activations to produce final prediction.

# step 6 Define Error and Loss

## Task 6.1 Error

```python
error = y_hat - y
# Error measures how far the prediction is from the true value
```

## Task 6.1 Compute Loss

```python
loss = np.mean(error**2)
# Mean Squared Error penalizes large mistakes more than small ones.
```

Q) why large errors matter more than small ones here

Ans. Squaring amplifies large mistakes, forcing the model to correct them more
aggressively.


# Step 7 Why we need slopes

To reduce loss, we must know how a small change in each parameter affects the loss. This information is given by the slope (gradient). Without slopes, we cannot update parameters correctly.

# Step 8 Backward Flow

## task 8.1 from loss to prediction

```python
dL_dy = 2 * error / len(x)
```

## Task 8.2 — Gradients of output layer

```python
dL_dW2 = h.T @ dL_dy
dL_db2 = np.sum(dL_dy, axis=0, keepdims=True)
```

Q) why these depend on hidden values


Ans. The output is computed using hidden activations.
So any change in W2 affects output through h

## Task 8.3 — Flow Into Hidden Layer

```python
# Gradient flowing into hidden layer
dL_dh = dL_dy @ w2.T
```

## Task 8.4 — Apply activation slope


```python
# Apply activation derivative
dL_dz1 = dL_dh * activation_slope(z1)
```

Q) How does activation control error flow?

Ans. Activation determines whether a neuron contributes to learning by allowing or blocking gradient flow.

## Task 8.5 — Gradients of first layer

```python
# Gradient for W1
dL_dW1 = x.T @ dL_dz1
# Gradient for b1
dL_db1 = np.sum(dL_dz1, axis=0, keepdims=True)
```

# Step 9 Parameter Update Rule

```python
# Learning rate
learning_rate = 0.01

# Update parameters
w1 -= learning_rate * dL_dW1
b1 -= learning_rate * dL_db1
w2 -= learning_rate * dL_dW2
b2 -= learning_rate * dL_db2
```

Q) Connection to Perceptron

Similar to perceptron learning, we adjust weights in the opposite direction of error to reduce mistakes.

```python
# ---------------------------------------------------------
# STEP 10 — TRAINING LOOP
# ---------------------------------------------------------

learning_rate = 0.01
epochs = 1000

# Justification:
# 1000 epochs allow gradual convergence.
# 0.01 learning rate keeps updates stable and prevents divergence.

for epoch in range(epochs):

    # -------- Forward Pass --------
    z1 = x @ w1 + b1
    h = activation(z1)
    y_hat = h @ w2 + b2
    # -------- Loss --------
    error = y_hat - y
    loss = np.mean(error ** 2)

    # -------- Backward Pass --------
    dL_dy = 2 * error / len(x)

    dL_dW2 = h.T @ dL_dy
    dL_db2 = np.sum(dL_dy, axis=0, keepdims=True)

    dL_dh = dL_dy @ w2.T
    dL_dz1 = dL_dh * activation_slope(z1)

    dL_dW1 = x.T @ dL_dz1
    dL_db1 = np.sum(dL_dz1, axis=0, keepdims=True)

    # -------- Update --------
    w1 -= learning_rate * dL_dW1
    b1 -= learning_rate * dL_db1
    w2 -= learning_rate * dL_dW2
    b2 -= learning_rate * dL_db2

    if epoch % 100 == 0:
        print(epoch, loss)
```
