```python
import pandas as pd
df=pd.read_csv("/content/multiple_linear_regression_dataset.csv")
print(df.head(5))
print(df.columns)
print(df.shape)
```

```python
#Q Which columns are inputs?
#Ans age,experience
#Q Which column is the output?
#Ans income
#Q How many features does your model need to handle?
#Ans 2
```

```python
# Dividing the input and output features
X=df[["age","experience"]]
y=df["income"]
```

```python
X.shape,y.shape
```

```python
#Q What is the shape of X ?
#Ans (20,2)
#Q What is the shape of y ?
#Ans (20,)
#Q Why does X have 2 columns but y only one?
#Ans Because X represent the input features and y represent the output features as stated above we have two input features and
# one output features, and these two input features are used to calculate the ouput feature thus making X having two columsn and y one column
```

```python
import numpy as np
n_features = X.shape[1]
w = np.zeros(n_features)
b=0.0
```

```python
#Q Why do we need one weight per feature?
#Ans we need one weight per column because each column will affect the output differently
#Q Why is bias separate?
#Ans bias gave us the offset
#Q Would initializing with large values be risky?
#Ans Intializing with larges values will not be riskty according to me but it may take long to converge in this according to me
```

```python
def predict(X,w,b):
  y_hat = X.dot(w) + b
  return y_hat
```

```python
#Q Why is there no activation function?
#Ans because we are predicting a numerical value
#Q What kind of values can y_hat take?
#Ans It could be any value ranging from negative to positive
#Q How is this different from logistic regression?
#Ans Logistic Regression work on classification type of the problem and gave us confidence level for it
```

```python
def meanSquared_error(y,y_hat):
  loss=((y_hat - y)**2).mean()
  return loss
```

```python
#Q Why square the error?
#Ans in order to penalize the large values more
#Q What happens if one prediction is very wrong?
#Ans It will affect the error if the number of the data row are very small but in the case of the large rows it will not have much effect
#Q Why not just take absolute error?
#Ans There is a problem for the gradient as at x = 0 the slope is nagtive at one side and positive at the other side
```

```python
def compute_gradients(X,y,y_hat):
  N=len(y)
  dw=(2/N)* X.T.dot(y_hat -y)
  db=(2/N)* (y_hat -y).sum()
  return dw,db
```

```python
#Q Why does X appear in dw but not in db ?
#Ans The weights are multiplied by the input values, so the gradient depends on 𝑋. The bias isn’t multiplied by anything, so X doesn’t appear in db.
#Q Why does the error term appear everywhere?
#Ans The error shows how wrong the prediction is. Gradients use it to know how much to change the weights and bias.
#Q What happens if error is zero?
#Ans If the error is zero, the prediction is perfect. Gradients become zero, so the weights and bias don’t change.
```

```python
def update_parameters(w,b,dw,db,lr):
  w= w- lr*dw
  b= b- lr*db
  return w,b
```

```python
lr =0.0001
epochs=1000
for epoch in range(epochs):
  y_hat=predict(X,w,b)
  loss = meanSquared_error(y,y_hat)
  dw,db=compute_gradients(X,y,y_hat)
  w,b = update_parameters(w,b,dw,db,lr)
  if(epoch % 100 == 0):
    print(f"Epoch {epoch}, Loss:{loss}")
```

```python
#Q Does loss decrease over time?
#Ans yes
#Q What happens if it increases?
#Ans if it increases , we must adjust the parameters such as learning rate and epochs
#Q How do learning rate and epochs interact?
#Ans learning rate determines te rate of learning of the model after each epoch
```

```python
print("final weight:\n",w)
print("\n")
print("final bias:\n",b)
```

```python
new_candidate =np.array([68,4.5])
predicted_salary=new_candidate.dot(w)+b
print(predicted_salary)
```
