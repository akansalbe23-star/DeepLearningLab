## Training a Perceptron


```python
import numpy as np
class Perceptron:
  def __init__(self,learning_rate=0.01,n_iters=1000):
    self.lr=learning_rate
    self.n_iters=n_iters
    self.activation_function = self._relu_function
    self.weights = None
    self.bias = None
  def fit(self,X,y):
    n_samples,n_features = X.shape
    self.weights = np.zeros(n_features)
    self.bias = 0

    for _ in range(self.n_iters):
      for idx,x_i in enumerate(X):
        linear_output = np.dot(x_i,self.weights) + self.bias
        y_predicted = self.activation_function(linear_output)

        # Perceptron udate rule
        update = self.lr * (y[idx] - y_predicted)
        self.weights += update * x_i
        self.bias += update
  def _relu_function(self,x):
    return np.maximum(0,x)
  def predict(self,X):
    linear_output = np.dot(X,self.weights) + self.bias
    y_predicted = self.activation_function(linear_output)
    return y_predicted
```

```python
X = np.array([[1,0],[0,1],[1,0],[1,1]]) # Logical And dataset
y = np.array([0,0,0,1])
```

```python
perceptron = Perceptron(learning_rate=0.1,n_iters=10)
perceptron.fit(X,y)
```

```python
print("Predictions : ",perceptron.predict(X))
```

## Activatin functions


```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
```

```python
# Generate Data Points
X = tf.linspace(-10,10,100)
X
```

```python
X.shape
```

```python
y = tf.keras.activations.sigmoid(X)
plt.plot(X,y,label='sigmoid = 1/(1+e^-x)')
plt.title('Sigmoid')
plt.xlabel('y')
plt.ylabel('Sigmoid of X')
plt.legend()
plt.show()
```

```python
# tanh
y = tf.keras.activations.tanh(X)
plt.plot(X,y,label='tanh(x)')
plt.title('Hyperbolic Tangent')
plt.xlabel('X')
plt.ylabel('tanh(X)')
plt.legend()
plt.show()
```

```python
# RELU
y = tf.keras.activations.relu(X)
plt.plot(X,y,label='y = max(0,x)')
plt.title('Relu')
plt.xlabel('X')
plt.ylabel('max(0,x)')
plt.legend()
plt.show()
```

```python
# Leaky RELU
y = tf.nn.leaky_relu(X)
plt.plot(X,y,label='leaky relu')
plt.title('Relu')
plt.xlabel('X')
plt.ylabel('leaky relu')
plt.legend()
plt.show()
```
